```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("temperatures_w1:", temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
fahrenheit = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", fahrenheit)
print("Max:", np.max(temperatures_w1))
print("Min:", np.min(temperatures_w1))
print("Mean:", np.mean(temperatures_w1))
```

```python
print("First 3 days:", temperatures_w1[0:3])
print("Weekend (last 2 days):", temperatures_w1[-2:])
print("Middle 3 days:", temperatures_w1[2:5])
```

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print(temperatures)
```

```python
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total elements:", temperatures.size)
print("Week 1:", temperatures[0, :])
print("Week 2:", temperatures[1, :])
print("Weekends for both weeks:", temperatures[:, -2:])
```

```python
import pandas as pd
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1','Rank2','Rank3','Rank4','Rank5'])
print(marks)
```

```python
print("\n1st rank (integer position):", marks[0])
print("\nTop 3 ranks using loc:\n", marks.loc['Rank1':'Rank3'])
print("\n3rd rank using iloc:", marks.iloc[2])
print("\nMarks > 90:\n", marks[marks > 90])
```

```python
marks['Rank1'] = 100
print("\nAfter modifying Rank1 to 100:\n", marks)

marks = marks.drop('Rank5')
print("\nAfter removing last rank:\n", marks)

cgpa = marks / 10
print("\nCGPA (marks/10):\n", cgpa)
```

```python
import pandas as pd
transactions = pd.DataFrame({
    'TransactionID': [101,102,103,104,105,106,107,108,109,110],
    'ProductCategory': ['Electronics','Clothing','Electronics','Furniture','Clothing','Electronics','Furniture','Clothing','Furniture','Electronics'],
    'Region': ['North','South','North','East','West','North','East','West','South','North'],
    'Amount': [200,150,300,450,200,250,300,180,350,400]
})
print(transactions)
```

```python
print("\nHead:\n", transactions.head())
print("\nTail:\n", transactions.tail())
print("\nShape:", transactions.shape)
print("\nColumns:", transactions.columns.tolist())
print("\nData types:\n", transactions.dtypes)

print("\nOnly ProductCategory and Amount:\n", transactions[['ProductCategory','Amount']])
print("\nLast 3 columns using iloc:\n", transactions.iloc[:, -3:])
print("\nFilter - North and Amount > 200:\n", transactions[(transactions['Region']=='North') & (transactions['Amount']>200)])
print("\nValue counts:\n", transactions['ProductCategory'].value_counts())
print("\nUnique Regions:", transactions['Region'].unique())
print("\nMean Amount by Region:\n", transactions.groupby('Region')['Amount'].mean())
```

```python
transactions.loc[transactions['TransactionID']==102, 'Amount'] = 165
print("\nAfter modifying 102 to 165:\n", transactions)

transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAfter adding Discount column:\n", transactions)

transactions = transactions[transactions['TransactionID']!=109]
print("\nAfter removing TransactionID 109:\n", transactions)

transactions = transactions.drop(columns=['Discount'])
print("\nAfter deleting Discount column:\n", transactions)
```
